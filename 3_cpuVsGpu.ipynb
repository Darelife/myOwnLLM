```python
import torch
import numpy as np
import time
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(device)
```

```python
%%time
start_time = time.time()
zeroes = torch.zeros(1,1, device=device)
end_time = time.time()

elapsed_time = end_time - start_time
print(f"Time taken to create a tensor of zeros: {elapsed_time:.8f}")
```

```python
dim = 100
number_of_dims = 4
tuple_dims = (dim,) * number_of_dims
torch_rand1 = torch.rand(tuple_dims, device=device)
torch_rand2 = torch.rand(tuple_dims, device=device)
np_rand1 = np.random.rand(*tuple_dims)
np_rand2 = np.random.rand(*tuple_dims)

start_time = time.time() 
rand = torch_rand1 @ torch_rand2
end_time = time.time()
elapsed_time = end_time - start_time
print(f"Time taken for torch matrix multiplication: {elapsed_time:.8f}")

start_time = time.time()
rand = np.multiply(np_rand1, np_rand2)
end_time = time.time()
elapsed_time = end_time - start_time
print(f"Time taken for numpy matrix multiplication: {elapsed_time:.8f}")
```
