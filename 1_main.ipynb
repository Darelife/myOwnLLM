```python
with open("wizardofoz.txt", 'r', encoding='utf-8') as f:
  text = f.read()

chars = sorted(set(text))
print(chars)
print(len(chars))
```

```python
string_to_int = { ch: i for i, ch in enumerate(chars) }
int_to_string = { i: ch for i, ch in enumerate(chars) }
encode = lambda s: [string_to_int[c] for c in s]
decode = lambda l: ''.join([int_to_string[i] for i in l])

# print(decode(encode("hello")))
# encoded_hello = torch.tensor(encode("hello"), dtype=torch.long)
# decoded_hello = decode(encoded_hello.tolist())
# print(encoded_hello)
# print(decoded_hello)
data = torch.tensor(encode(text), dtype=torch.long)
print(data[:100])
```

```python
n = int(0.8*len(data))
trainData = data[:n]
testData = data[n:]
```

# Block Size

So, what it means is that we are going to look at the previous 5 characters to predict the next character.

X is the input and Y is the target. So, we are going to look at the previous 5 characters to predict the next character.

In the code below, we've defined a block size of 5, and we are going to look at the first 5 characters of the training data as input and the next character as the target. We will then loop through the block size and print out the input and target for each step.

Here, we aren't really predicting anything yet, we are just looking at the data and understanding how it is structured. We will be using this structure to train our model later on.

```python
x = trainData[:blockSize]
y = trainData[1:blockSize+1]

for t in range(blockSize):
  context = x[:t+1]
  target = y[t]
  print(f"when input is {context.tolist()} the target: {target}")
```
